```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el archivo JSON
file_path = "personas_encuesta_cpi_mayo2025.json"
with open(file_path, "r", encoding="utf-8") as f:
    personas = json.load(f)

# 2. Convertir a DataFrame
df = pd.DataFrame(personas)

# 3. Configurar estilo visual
sns.set(style="whitegrid")

# 4. Función para graficar porcentajes
def plot_percentage_bar(column):
    plt.figure(figsize=(10, 5))
    value_counts = df[column].value_counts(normalize=True).sort_values(ascending=False)
    ax = sns.barplot(x=value_counts.index, y=value_counts.values * 100)
    plt.title(f'Distribución porcentual de: {column}', fontsize=14)
    plt.ylabel('Porcentaje (%)')
    plt.xticks(rotation=45, ha='right')
    for i, v in enumerate(value_counts.values):
        ax.text(i, v * 100 + 1, f"{v * 100:.1f}%", ha='center')
    plt.tight_layout()
    plt.show()

# 5. Graficar todas las columnas
for col in df.columns:
    plot_percentage_bar(col)
```
